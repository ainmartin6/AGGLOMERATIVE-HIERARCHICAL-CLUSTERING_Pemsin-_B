```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
```

```python
df = pd.read_csv("health.csv" , sep = ';')
df
```

```python
df = np.array(df[['User_Satisfaction_Rating', 'Performance_Score']])
df
```

```python
plt.title("Dendogram Method Single Linkage")
dendogram = sch.dendrogram(sch.linkage(df, method = 'single')) 
```

```python
plt.title("Dendogram Method Average Linkage")
dendogram = sch.dendrogram(sch.linkage(df, method = 'average')) 
```

```python
plt.title("Dendogram Method Complate Linkage")
dendogram = sch.dendrogram(sch.linkage(df, method = 'complete')) 
```

```python
 plt.title("Dendogram Method Ward Linkage")
dendogram = sch.dendrogram(sch.linkage(df, method = 'ward')) 
```

```python
from sklearn.cluster import AgglomerativeClustering as acg
```

```python
ac = acg(n_clusters = 4, linkage = 'single')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter Plot')
plt.xlabel('Customer Satisfaction')
plt.ylabel('Performance Score')
plt.show()
```

```python
ac = acg(n_clusters = 4, linkage = 'average')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter average linkage Plot')
plt.xlabel('Customer Satisfaction')
plt.ylabel('Performance Score')
plt.show() 
```

```python
ac = acg(n_clusters = 4, linkage = 'complete')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter complate linkage Plot')
plt.xlabel('Customer Satisfaction')
plt.ylabel('Performance Score')
plt.show() 
```

```python
ac = acg(n_clusters = 4, linkage = 'ward')
output_eu = ac.fit_predict(df)

plt.scatter(df[:,0], df[:,1], c = output_eu, s = 50, cmap = 'rainbow')
plt.title('Scatter ward linkage Plot')
plt.xlabel('Customer Satisfaction')
plt.ylabel('Performance Score')
plt.show() 
```
